# 定义输出格式和提示词


```json
{
  "result": [
    {
      "speaker": "中文姓名",
      "content": "发言内容",
      "is_fraud": true,
      "reason": "如果 is_fraud=true，说明原因；否则为空字符串"
    }
  ]
}
```

````python
from typing import List

from langchain_core.output_parsers import JsonOutputParser
from langchain_core.prompts import ChatPromptTemplate
from pydantic import BaseModel, Field

# 约定LLM的输出应该是什么样子的， 这里Field是说明，用来给LLM看的
class DialogItem(BaseModel):
    speaker: str = Field(description="为发言人生成的中文姓名，不含角色名")
    content: str = Field(description="发言的详细内容")
    is_fraud: bool = Field(description="这句发言中是否存在明确具体、不可辩驳的欺诈内容")
    reason: str = Field(description="仅当 is_fraud=true 时给出理由；否则为空字符串")

# 这里让LLM返回一个List的形式
class DialogResult(BaseModel):
    result: List[DialogItem]
    
# 这里会被jsonoutputParser 从字符串转化成这些json
parser = JsonOutputParser(pydantic_object=DialogResult)

prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "你是一个分析诈骗案例并且擅长写作的专家。你的任务是将诈骗的案发过程以口语对话的形式完整还原。",
    ),
    (
        "human",
        '''
请根据下面的诈骗案例报道生成对话数据。

要求：
1. 分析案例中描述的案发过程，识别诈骗者和受害者，并为每个角色生成一个中文姓名。
2. 按照案发前后经过，将过程改写成诈骗者和受害者之间循序渐进的中文口语对话。
3. 诈骗者的发言要尽量符合真实诈骗话术：看起来正规、可信、能降低受害者警惕。
4. 从第一句发言开始，逐句判断该发言是否存在明确具体、不可辩驳的欺诈内容。
5. 只输出 JSON，不要输出 Markdown 代码块，不要输出额外解释。

案例报道：
{text}

输出格式说明：
{format_instructions}
''',
    ),
])

chain = prompt | llm | parser
````

````python
def remove_markdown_boundary(text):
    """兼容某些模型喜欢返回 ```json ... ``` 的情况。"""
    return re.sub(r"^```(?:json)?\s*(.*?)\s*```$", r"\1", text.strip(), flags=re.DOTALL)


def normalize_dialog_response(response):
    """把 LangChain/parser 或原始字符串的返回统一成 result 列表。"""
    if isinstance(response, str):
        response = json.loads(remove_markdown_boundary(response))

    if isinstance(response, dict) and "result" in response:
        return response["result"]

    if isinstance(response, list):
        return response

    raise ValueError(f"无法识别模型返回格式：{response}")


def generate_dialog(text):
    response = chain.invoke({
        "text": text,
        "format_instructions": parser.get_format_instructions(), # format instruction 直接填写到前面prompt里面
    })
    return normalize_dialog_response(response)
````

````python
def filename(path):
    return Path(path).stem


def build_write_func(file_name):
    file_name = Path(file_name)

    def write(rows, header=False):
        df_new = pd.DataFrame(rows, index=range(len(rows)))
        df_new.to_csv(file_name, mode="a", header=header, index=False, encoding="utf-8-sig")

    return write


def build_to_rows_func(case_prefix=""):
    def to_rows(dialog, i):
        rows = []
        for item in dialog:
            rows.append({
                "case": f"{case_prefix}{i}",
                "speaker": item.get("speaker", ""),
                "content": item.get("content", ""),
                "is_fraud": item.get("is_fraud", False),
                "reason": item.get("reason", ""),
            })
        return rows

    return to_rows
````

````python
def generate_dialogs(input_path, output_dir, start_idx=0, end_idx=None, overwrite=False):
    input_path = Path(input_path)
    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)

    dataset = pd.read_csv(input_path)
    output_path = output_dir / f"{filename(input_path)}.csv"

    if overwrite and output_path.exists():
        output_path.unlink()

    write_fn = build_write_func(output_path)
    to_rows_fn = build_to_rows_func(filename(input_path))

    if end_idx is None:
        rows_iter = dataset.iloc[start_idx:].iterrows()
    else:
        rows_iter = dataset.iloc[start_idx:end_idx].iterrows()

    wrote_header = output_path.exists() and output_path.stat().st_size > 0

    for i, row in rows_iter:
        try:
            dialog = generate_dialog(row[column_content])
            dialog_rows = to_rows_fn(dialog, i)
            write_fn(dialog_rows, header=not wrote_header)
            wrote_header = True
            print(f"写入成功：{input_path.name} 第 {i} 条，共 {len(dialog_rows)} 句")
        except Exception as e:
            print(f"生成失败：{input_path.name} 第 {i} 条，错误：{e}")

    return output_path
````

````python
# 小批量测试：先只跑 2 条，确认格式没问题后再扩大范围
output_dir = Path("../datasets/fraud/csv_dialogs_langchain")

input_path = dataset_path / "冒充客服诈骗(200).csv"
output_path = generate_dialogs(input_path, output_dir, start_idx=0, overwrite=True)

pd.read_csv(output_path).head()
````
